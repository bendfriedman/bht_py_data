# Matplotlib: Die Basis

## Grundkonzepte

- **Figure**: Das gesamte Fenster/Bild
- **Axes**: Der Plot-Bereich (kann mehrere in einer Figure haben)
- **Axis**: X-Achse und Y-Achse

```python
import matplotlib.pyplot as plt

# Einfacher Plot
plt.plot([1, 2, 3, 4], [1, 4, 9, 16])
plt.xlabel('X-Achse')
plt.ylabel('Y-Achse')
plt.title('Mein erster Plot')
plt.show()
```

# Matplotlib: Airbnb Beispiel

```python
import pandas as pd
import matplotlib.pyplot as plt

# Daten laden
df = pd.read_csv('listings.csv')

# Preisverteilung als Histogramm
plt.figure(figsize=(10, 6))
plt.hist(df['price'], bins=50, edgecolor='black')
plt.xlabel('Preis (€)')
plt.ylabel('Anzahl')
plt.title('Preisverteilung der Airbnb Listings')
plt.grid(True, alpha=0.3)
plt.show()
```

# Matplotlib: Subplots

## Mehrere Plots in einer Figure

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

# Plot 1: Histogramm
axes[0, 0].hist(df['price'], bins=30)
axes[0, 0].set_title('Preisverteilung')

# Plot 2: Scatterplot
axes[0, 1].scatter(df['bedrooms'], df['price'])
axes[0, 1].set_title('Zimmer vs. Preis')

# Plot 3 & 4: ...

plt.tight_layout()
plt.show()
```

# Seaborn: Statistische Plots

## Einfachere Syntax, schönere Defaults

```python
import seaborn as sns

# Style setzen
sns.set_style('whitegrid')
sns.set_palette('husl')

# Boxplot nach room_type
plt.figure(figsize=(10, 6))
sns.boxplot(data=df, x='room_type', y='price')
plt.title('Preisverteilung nach Raumtyp')
plt.xticks(rotation=45)
plt.show()
```

# Seaborn: Airbnb Beispiele

## Verschiedene Plot-Typen

```python
# 1. Scatterplot mit Hue
sns.scatterplot(data=df, x='bedrooms', y='price', 
                hue='room_type', alpha=0.6)

# 2. Violin Plot
sns.violinplot(data=df, x='neighbourhood_group', 
               y='price')

# 3. Count Plot
sns.countplot(data=df, x='room_type')
```

# Seaborn: Heatmap für Korrelationen

```python
# Numerische Spalten auswählen
numeric_cols = ['price', 'minimum_nights', 
                'number_of_reviews', 'reviews_per_month',
                'availability_365']

# Korrelationsmatrix berechnen
corr = df[numeric_cols].corr()

# Heatmap plotten
plt.figure(figsize=(10, 8))
sns.heatmap(corr, annot=True, cmap='coolwarm', 
            center=0, fmt='.2f')
plt.title('Korrelationsmatrix')
plt.show()
```

# Plotly: Interaktive Visualisierungen

## Warum Plotly?

- **Interaktivität**: Zoom, Pan, Hover
- **Web-ready**: Perfekt für Streamlit
- **3D-Plots**: Möglich!

```python
import plotly.express as px

# Interaktiver Scatterplot
fig = px.scatter(df, x='bedrooms', y='price',
                 color='room_type',
                 hover_data=['neighbourhood_group'],
                 title='Airbnb Preise')
fig.show()
```

# Plotly: Airbnb Beispiele

## Verschiedene interaktive Plots

```python
# 1. Histogramm
fig = px.histogram(df, x='price', nbins=50,
                   title='Preisverteilung')

# 2. Box Plot
fig = px.box(df, x='room_type', y='price',
             title='Preis nach Raumtyp')

# 3. Bar Chart
room_counts = df['room_type'].value_counts()
fig = px.bar(x=room_counts.index, y=room_counts.values,
             title='Anzahl Listings pro Raumtyp')
```

# Plotly: Geografische Visualisierung

```python
# Scatter Map für Airbnb Locations
fig = px.scatter_mapbox(
    df,
    lat='latitude',
    lon='longitude',
    color='price',
    size='price',
    hover_name='name',
    hover_data=['room_type', 'price'],
    color_continuous_scale='Viridis',
    zoom=10,
    height=600
)

fig.update_layout(mapbox_style='open-street-map')
fig.show()
```

# Best Practices für gute Visualisierungen

## ✓ Dos
- Klare, beschreibende Titel und Labels
- Passende Farben wählen (Colorblind-friendly)
- Achsenbeschriftungen nicht vergessen
- Legende hinzufügen (wenn mehrere Kategorien)
- Angemessene Größe der Figure

## ✗ Don'ts
- 3D-Plots ohne Mehrwert
- Zu viele Farben oder Informationen
- Unleserliche Schriftgrößen
- Fehlende Einheiten bei Achsen

# Library-Vergleich: Wann was nutzen?

```python
# Matplotlib: Volle Kontrolle, komplex
fig, ax = plt.subplots()
ax.scatter(df['x'], df['y'])
ax.set_xlabel('X')
ax.set_ylabel('Y')

# Seaborn: Statistisch, einfach, schön
sns.scatterplot(data=df, x='x', y='y', hue='category')

# Plotly: Interaktiv, web-ready
px.scatter(df, x='x', y='y', color='category')
```
