# 2. Fehlende Werte (Missing Data)

## Identifizieren

```python
# Fehlende Werte identifizieren
df.isnull().sum()
df.isna().sum()

# Visualisieren
import missingno as msno
msno.matrix(df)
```

## Drei Hauptansätze

| Methode | Funktion | Wann? |
|---------|----------|-------|
| Löschen | `dropna()` | Viele fehlende Werte |
| Ersetzen | `fillna()` | Einzelne fehlende Werte |
| Interpolieren | `interpolate()` | Zeitreihen |

# Strategien für Fehlende Werte

## 1. Zeilen/Spalten entfernen
- Wenn viele (oder nicht kritische) Werte fehlen
- `df.dropna()`

## 2. Mit Konstante ersetzen
- Mittelwert, Median, Modus oder 0
- `df.fillna(df.mean())`

## 3. Forward/Backward Fill
- Zeitreihen: vorherigen/nächsten Wert nutzen
- `df.fillna(method='ffill')`

## 4. Interpolation
- Werte zwischen bekannten Punkten schätzen
- `df.interpolate()`

# 3. Duplikate Erkennen & Entfernen

```python
# Duplikate finden
df.duplicated().sum()

# Duplikate anzeigen
df[df.duplicated()]

# Duplikate entfernen
df.drop_duplicates()

# Nur bestimmte Spalten prüfen
df.drop_duplicates(subset=['name'])
```

**Achtung:** Parameter `keep='first'`, `'last'` oder `False`

# 4. Ausreißer (Outliers)

## Erkennungsmethoden
- IQR-Methode (Interquartile Range)
- Z-Score
- Visualisierung (Boxplot) --> sehr wichtig!

## Umgang mit Ausreißern
- Entfernen
- Winsorizing
- Transformation

```python
# IQR-Methode
Q1 = df['age'].quantile(0.25)
Q3 = df['age'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
```

```python
# Strings bereinigen
df['name'] = df['name'].str.strip()
df['name'] = df['name'].str.lower()
df['name'] = df['name'].str.title()

# Kategorien vereinheitlichen
df['sex'].replace({
    'M': 'male', 'm': 'male',
    'F': 'female', 'f': 'female'
}, inplace=True)
```
