# Warum NumPy?

Wer Daten analysiert, arbeitet viel mit **Zahlen und Matrizen**.

**NumPy** (Harris et al., 2020) steht für *numerical Python*: eine Bibliothek für wissenschaftliches Rechnen mit

- Arrays, Vektor- und Matrixberechnungen
- mathematischen Funktionen und Zufallszahlengeneratoren

Gründe für den Einsatz: **Komfort** (vor allem bei Matrixoperationen) und **Geschwindigkeit**. Viele Bibliotheken (z. B. `scipy`, `pandas`) haben die NumPy-API übernommen und erweitert.

# NumPy ist schnell

Erst die Python-Liste, dann das NumPy-Array: dieselbe Aufgabe (alle Werte verdoppeln) bei einer Million Elementen.

````python
my_list = list(range(1000000))
%timeit [x * 2 for x in my_list]
````

````python
import numpy as np

np.random.seed(0)   # Seed für Reproduzierbarkeit
my_arr = np.arange(1000000)
%timeit my_arr * 2
````

Die genauen Zeiten hängen vom Rechner ab. Typisch ist ein Faktor von **grob 50 bis 100**: von einigen zehn Millisekunden auf unter eine Millisekunde.

# Warum ist NumPy schneller als Listen?

Eine Python-Liste kann **beliebige Typen** enthalten:

````python
L3 = [True, "2", 3.0, 4]
[type(item) for item in L3]
````

Das kann **ineffizient** sein: Jedes Element ist ein eigenes Objekt mit Typinformation, die Liste speichert nur Zeiger darauf.

Ein NumPy-Array hat dagegen **einen einzigen Datentyp** und legt die Werte **hintereinander im Speicher** ab. Berechnungen laufen dann in schnellem, kompiliertem C-Code.

# Array-Datentypen

| Datentyp | Beschreibung |
|----------|--------------|
| `bool_` | Boolean (`True` oder `False`), als ein Byte gespeichert (ab NumPy 2 auch `np.bool`) |
| `int8`, `int16`, `int32`, `int64` | Ganzzahlen mit 8, 16, 32 oder 64 Bit, z. B. `int8`: −128 bis 127 (`int_`: Standard-Integer, auf 64-Bit-Systemen `int64`) |
| `uint8`, `uint16`, `uint32`, `uint64` | Vorzeichenlose Ganzzahlen, z. B. `uint8`: 0 bis 255 |
| `float16`, `float32`, `float64` | Gleitkommazahlen mit halber, einfacher und doppelter Genauigkeit (`float64` ist der Standard) |
| `complex64`, `complex128` | Komplexe Zahlen: für Data Scientists meist weniger relevant |

*Strukturierte Arrays mit zusammengesetzten Typen speichert man besser in einem **pandas DataFrame**. Der Alias `float_` wurde in NumPy 2 entfernt: stattdessen `float64` verwenden.*

# Array-Attribute

Wir erzeugen drei zufällige Arrays mit 1, 2 und 3 Dimensionen:

````python
x1 = np.random.randint(10, size=6)           # eindimensionales Array
x2 = np.random.randint(10, size=(3, 4))      # zweidimensionales Array
x3 = np.random.randint(10, size=(3, 4, 5))   # dreidimensionales Array
````

# Grundlegende Attribute

- `ndim`: Anzahl der Dimensionen
- `shape`: Größe jeder Dimension
- `size`: Gesamtgröße (Anzahl der Elemente)
- `dtype`: Datentyp der Elemente
- `itemsize`: Größe jedes Elements in Bytes
- `nbytes`: Größe des gesamten Arrays in Bytes

````python
print("x3 ndim:    ", x3.ndim)
print("x3 shape:   ", x3.shape)
print("x3 size:    ", x3.size)
print("x3 dtype:   ", x3.dtype)
print("x3 itemsize:", x3.itemsize)
print("x3 nbytes:  ", x3.nbytes)
````

# Masken kombinieren

Mehrere Bedingungen verknüpft man mit `&` (und), `|` (oder) und `~` (nicht). Jede Bedingung muss **eingeklammert** werden:

````python
grid[(grid > 2) & (grid < 7)]
````

````python
grid[(grid < 3) | (grid > 7)]
````

````python
grid[~(grid > 5)]
````

# Arithmetische Operationen

## Wiederholte Ausführung ist in Python langsam

Eine Schleife über jedes Element:

````python
def compute_reciprocals(values):
    output = np.empty(len(values))
    for i in range(len(values)):
        output[i] = 1.0 / values[i]
    return output


values = np.random.randint(1, 10, size=5)
compute_reciprocals(values)
````

````python
big_array = np.random.randint(1, 100, size=1000000)
%timeit compute_reciprocals(big_array)
````

Für eine Million Elemente dauert das **ein bis zwei Sekunden**.

# NumPy: Universal Functions (UFuncs)

NumPy ist schnell, weil Operationen über die **Universal Functions (ufuncs)** **vektorisiert** werden.

Für viele Aufgaben genügt **reines Python plus NumPy**, um effizient zu rechnen.

````python
1.0 / values   # vektorisierte, schnelle Kehrwertberechnung mit ufuncs
````

````python
%timeit (1.0 / big_array)
````

Dieselbe Berechnung: **grob 1000-mal schneller** als die Schleife (je nach Rechner).

# ufunc: Arithmetische Operatoren

| Operator | Äquivalente ufunc | Beschreibung |
|----------|-------------------|--------------|
| `+` | `np.add` | Addition (z. B. `1 + 1 = 2`) |
| `-` | `np.subtract` | Subtraktion (z. B. `3 - 2 = 1`) |
| `-` | `np.negative` | Unäre Negation (z. B. `-2`) |
| `*` | `np.multiply` | Multiplikation (z. B. `2 * 3 = 6`) |
| `/` | `np.divide` | Division (z. B. `3 / 2 = 1.5`) |
| `//` | `np.floor_divide` | Ganzzahldivision (z. B. `3 // 2 = 1`) |
| `**` | `np.power` | Potenzierung (z. B. `2 ** 3 = 8`) |
| `%` | `np.mod` | Modulo/Rest (z. B. `9 % 4 = 1`) |

# Operatoren auf Arrays

````python
a = np.array([1, 2, 3, 4])
print(a + 10)
print(a * 2)
print(a ** 2)
print(a + np.array([10, 20, 30, 40]))   # elementweise
````

# Aggregationsfunktionen

Bei großen Datensätzen helfen Aggregationen, die Daten zu verstehen:

- Summe aller Werte
- Minimum/Maximum
- Quantile
- Mittelwert/Median

NumPy bietet dafür schnelle, vektorisierte Funktionen.

````python
L = np.arange(10)
L
````

````python
print(sum(L))      # eingebaute Python-Funktion
print(np.sum(L))   # NumPy-Funktion
````

# Aggregationsfunktionen in NumPy

| Funktion | Beschreibung |
|----------|--------------|
| `np.sum`, `np.prod` | Summe, Produkt der Elemente |
| `np.mean`, `np.median` | Mittelwert, Median |
| `np.std`, `np.var` | Standardabweichung, Varianz |
| `np.min`, `np.max` | Minimum, Maximum |
| `np.argmin`, `np.argmax` | Index des Minimums, des Maximums |
| `np.percentile` | Rangbasierte Statistiken (Perzentile) |
| `np.any`, `np.all` | Ist mindestens ein / sind alle Elemente wahr? |

Zu fast allen gibt es eine **NaN-sichere Version** mit Präfix `nan`, z. B. `np.nansum` oder `np.nanmean` (nicht für `any` und `all`).

# NaN-sichere Funktionen

`NaN` (*not a number*) steht für fehlende oder undefinierte Werte. Normale Aggregationen liefern damit `nan`, die `nan`-Versionen ignorieren die fehlenden Werte:

````python
data = np.array([1.0, np.nan, 3.0])
print(data)
print(np.sum(data))
print(np.nansum(data))
print(np.nanmean(data))
````
