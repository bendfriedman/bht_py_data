```python
def create_grid(rows: int, cols: int) -> np.ndarray:
    """Gibt ein (rows, cols)-Array mit den Zahlen 1, 2, 3, ... zurück."""
    # TODO
    return np.arange(1, rows * cols + 1).reshape((rows,cols))
```

### Aufgabe A5: Boolean Indexing

Schreibt `multiples_above(arr, k, threshold)`: Die Funktion gibt ein Array mit allen Elementen von `arr` zurück, die **durch `k` teilbar** und **größer als `threshold`** sind. Verwendet eine **Boolean-Maske** (mit `&` kombiniert, jede Bedingung in Klammern).

Beispiel: `multiples_above(np.arange(1, 21), 3, 8)` → `[9, 12, 15, 18]`

```python
def multiples_above(arr: np.ndarray, k: int, threshold: int) -> np.ndarray:
    """Gibt alle Elemente zurück, die durch k teilbar und größer als threshold sind."""
    # TODO
    print("arr: ", arr, " k: ", k, " threshold: ", threshold)
    result = arr[(arr % k == 0) & (arr > threshold)]
    print("result: ", result)
    return result
```

```python
numbers = np.arange(1, 21)
assert multiples_above(numbers, 3, 8).tolist() == [9, 12, 15, 18], "Beispiel aus der Aufgabe"
assert numbers.tolist() == list(range(1, 21)), "Das Original darf nicht verändert werden"
assert multiples_above(numbers, 5, 100).tolist() == [], "Keine Treffer"
assert multiples_above(np.array([4, 8, 12]), 4, 0).tolist() == [4, 8, 12], "Alle Treffer"
print("✓ Aufgabe A5 bestanden")
```

### Aufgabe B1: Standardisieren ohne Schleife

Schreibt `standardize(arr)`: Die Funktion gibt ein **neues** Array zurück, in dem von jedem Wert der **Mittelwert** abgezogen und das Ergebnis durch die **Standardabweichung** geteilt wird (`np.mean`, `np.std`). Verwendet **keine Schleife**. Ist die Standardabweichung `0` (alle Werte gleich), besteht das Ergebnis aus **lauter Nullen**.

Das Ergebnis hat Mittelwert `0` und Standardabweichung `1`.

```python
def standardize(arr: np.ndarray) -> np.ndarray:
    """Gibt (arr - Mittelwert) / Standardabweichung zurück (Nullen bei Standardabweichung 0)."""
    # TODO
    print("array: ", arr)
    if arr.std() == 0:
        return np.zeros(arr.shape, dtype=float)
    result = (arr - arr.mean()) / arr.std()
    print("result: ", result)
    return result.copy()
```

```python
data = np.array([2.0, 4.0, 4.0, 4.0, 5.0, 5.0, 7.0, 9.0])
result = standardize(data)
assert result.shape == data.shape, "Die Form soll erhalten bleiben"
assert abs(np.mean(result)) < 1e-9, f"Mittelwert: {np.mean(result)}"
assert abs(np.std(result) - 1) < 1e-9, f"Standardabweichung: {np.std(result)}"
assert np.allclose(result[:2], [-1.5, -0.5]), "Konkrete Werte"
assert data.tolist() == [2.0, 4.0, 4.0, 4.0, 5.0, 5.0, 7.0, 9.0], "Das Original darf nicht verändert werden"
assert standardize(np.array([3.0, 3.0, 3.0])).tolist() == [0.0, 0.0, 0.0], "Standardabweichung 0"
print("✓ Aufgabe B1 bestanden")
```

### Aufgabe B2: Schleife oder Vektorisierung?

Berechnet die **Summe der Quadrate** der Zahlen `0` bis `n - 1` auf zwei Wegen:

- `sum_of_squares_loop(n)`: mit einer normalen **Python-Schleife** (`for` über `range(n)`).
- `sum_of_squares_vec(n)`: **vektorisiert** mit NumPy, ohne Schleife. Erzeugt das Array mit `np.arange(n, dtype=np.int64)`, damit auch bei großen `n` nichts überläuft.

Beide Funktionen geben dieselbe ganze Zahl zurück. Vergleicht danach in der **Zeitmessungs-Zelle** unter dem Test die Laufzeiten mit `%timeit`.

Beispiel: `n = 4` → `0 + 1 + 4 + 9 = 14`

```python
def sum_of_squares_loop(n: int) -> int:
    """Summe der Quadrate von 0 bis n-1 mit einer Python-Schleife."""
    # TODO
    result = 0
    for x in range(n):
        result += x**2
    return int(result)
        


def sum_of_squares_vec(n: int) -> int:
    """Summe der Quadrate von 0 bis n-1 vektorisiert mit NumPy."""
    # TODO
    return int(np.sum((np.arange(n, dtype=np.int64)) ** 2))
```

```python
assert sum_of_squares_loop(4) == 14, "Schleife: Beispiel aus der Aufgabe"
assert sum_of_squares_vec(4) == 14, "Vektorisiert: Beispiel aus der Aufgabe"
assert sum_of_squares_loop(0) == 0 and sum_of_squares_vec(0) == 0, "n = 0"
for n in (1, 10, 1000):
    assert sum_of_squares_loop(n) == sum_of_squares_vec(n), f"Die Ergebnisse unterscheiden sich bei n = {n}"
assert sum_of_squares_vec(1_000_000) == 333332833333500000, "Große Zahlen (int64 verwenden)"
print("✓ Aufgabe B2 bestanden")
```
