# Das Modul auf einen Blick

- **Data Science Programmierung mit Python** (DataPy), 5. Semester Bachelor Informatik
- **13 Wochen**, wöchentlich **mittwochs**: Vorlesung mit Live-Coding und Übung
- von den Python-Grundlagen bis zur Projektpräsentation
- Materialien in **Moodle**: Jupyter Notebooks (Folien) und Bilder

# Kernziele

Nach dem Modul könnt ihr:

1. **Python** sicher anwenden (Datentypen, Kontrollstrukturen, Funktionen, Klassen)
2. mit **NumPy** und **pandas** Daten einlesen, auswählen und aggregieren
3. Daten aus **Dateien, Web und APIs** beschaffen
4. Daten **bereinigen, transformieren** und **visualisieren**
5. optional **ML-Modelle** (scikit-learn) trainieren und bewerten
6. Ergebnisse als **Streamlit-App** präsentieren
7. ein **Datenprojekt** von der Idee bis zur Präsentation durchführen

# So läuft eine Woche ab

1. **Vorlesung** mit Live-Coding: Folien sind ein Jupyter Notebook
2. **Übung:** kurze Aufgaben, oft am Airbnb-Datensatz
3. **Projektschritt:** ihr wendet das Thema auf **euren Datensatz** an
4. **Abgabe** über Moodle (siehe Wochenübersicht)

**Tipp:** Notebook vorab öffnen, mitmachen und eigene Notizen ergänzen

# Das Semesterprojekt

- **Individuell**, mit **eigenem Datensatz** und eigener Fragestellung
- Ab **Woche 6**: Idee und Datensatz, dann **jede Woche ein Baustein**
- Ergebnis: **Streamlit-App** mit Filtern und Visualisierungen
- **Machine Learning** ist optional
- Das Herzkrankheiten-Beispiel dient als **Musterlösung**, ist aber als Projektdatensatz **nicht zulässig**
- Abschluss: **Präsentation** und Feedbackrunde

# Modulbewertung

- **keine Prüfung** 
- ein **Semesterprojekt** mit Dokumentation und Präsentation sowie **2 Moodle-Tests**
- Mündliches **Gespräch zum Projekt**: ihr erklärt eure Entscheidungen (Bereinigung, Methoden, Bewertung der Ergebnisse)
- **Modulendnote** 50% Test (Mittelwert) + 50% Projekt (Abgabe+Präsentation)
