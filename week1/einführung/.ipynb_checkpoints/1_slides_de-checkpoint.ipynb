# Java vs. Python

## Drei Unterschiede

1. Kompilieren vs. Interpretieren
2. Eine Hauptversion vs. mehrere Virtual Environments
3. Packages und Module

# Compile vs. Interpret

<img src="images/compile_vs_interpret.png" alt="Ablauf bei Java (Compiler, Bytecode, JVM) im Vergleich zu Python (Bytecode, virtuelle Maschine)" style="display:block; margin:0 auto; width:100%;">

# Compile vs. Interpret

## Java

- Ein Java-Programm wird **kompiliert**
- Ausführung erst möglich, wenn das **gesamte** Programm erfolgreich kompiliert wurde

## Python

- Konzeptionell **interpretiert**: Der Code wird „direkt“ Zeile für Zeile ausgeführt, bei einem Fehler hält das Programm an
- Technisch etwas komplizierter (im nicht-interaktiven Modus):
  - Das Programm wird zuerst in **Bytecode** kompiliert
  - Eine **virtuelle Maschine** führt den Bytecode Zeile für Zeile aus und hält bei einem Fehler an

# Compile vs. Interpret: Beispiel

**Laufzeitfehler** – alles vor der fehlerhaften Zeile wurde bereits ausgeführt:

```python
print("Zeile 1")
print("Zeile 2")
print(zahl)          # zahl wurde nie definiert
print("Zeile 4")
```

```text
Zeile 1
Zeile 2
NameError: name 'zahl' is not defined
```

**Syntaxfehler** – werden schon beim Kompilieren erkannt, **vor** der Ausführung:
Von einem Skript mit einer fehlenden Klammer wird *keine* Zeile ausgeführt (`SyntaxError`).

# Hauptversion vs. Virtual Environments

<img src="images/multiple_venvs.png" alt="Java: ein JDK für alle Programme; Python: mehrere Virtual Environments mit eigenen Versionen" style="display:block; margin:0 auto; width:100%;">

# Hauptversion vs. Virtual Environments

## Java

- Eine (primäre) Version ist im Rechner installiert
- Alle Programme inklusive IDEs (Eclipse, IntelliJ, etc.) nutzen diese Version

## Python

- In **einer** Virtual Environment gibt es **eine** bestimmte Python-Version samt bestimmten Versionen der Libraries
- Mehrere Virtual Environments nebeneinander sind möglich

# Packages und Module

<img src="images/packages_modules.png" alt="Python-Package als Ordner mit __init__.py und Modulen als Dateien" style="display:block; margin:0 auto; width:100%;">

# Packages und Module

## Java

- Klasse ≈ Datei

## Python

- **Package** ≈ Ordner
- **Modul** ≈ Datei mit 0, 1 oder mehreren Klassen
- `__init__.py` wird beim Import ausgeführt

# Virtual Environments

- Für verschiedene Projekte brauchst du unterschiedliche Abhängigkeiten (oder Versionen davon)
- Best Practice: Abhängigkeiten in einer Virtual Environment kapseln

Steht der Name einer Virtual Environment in Klammern am Anfang der Eingabezeile, ist sie aktiviert:

```text
(base) C:\Users\Max>
```

Hier ist die Environment `base` aktiv (bei Anaconda die Standard-Environment).

# Python-Code ausführen: vier Methoden

**Nicht interaktiv**

1. Gesamtes Programm per Shell-Befehl ausführen: `python datei.py`

**Interaktiv**

2. Python-Interpreter: `python`
3. IPython-Interpreter: `ipython`
4. Jupyter Notebook (IPython-Interpreter, im Browser bereitgestellt)

Pros und Cons besprechen wir, nachdem wir alle vier Methoden angesehen haben.

# Methode 1: Skript per Shell ausführen

Datei `hello.py`:

```python
print("hello world")
```

Aufruf in der Shell:

```text
$ python hello.py
hello world
```

# Methode 2: Python-Interpreter

`python` in der Eingabeaufforderung / im Terminal eingeben:

```text
$ python
>>>
```

Code-Schnipsel eingeben und direkt ausführen:

```python
>>> 1 + 1
2
>>> x = 5
>>> x * 3
15
```

Eine Code-Datei ausführen:

```python
exec(open('hello.py').read())
```

# Methode 3: IPython-Interpreter

IPython erweitert die Python-Shell um viele nützliche Funktionen:

- Tab-Completion
- Verlauf (History)
- Hilfe / Dokumentation
- Plotting

`ipython` in der Eingabeaufforderung / im Terminal eingeben. Der sichtbare Unterschied:

- Python: `>>>`
- IPython: nummerierte Eingaben (`In [1]:`) und Ausgaben (`Out[1]:`)

# Methode 3: IPython-Interpreter – Beispiel

```python
In [1]: 1 + 1
Out[1]: 2

In [2]: x = 5

In [3]: x * 3
Out[3]: 15
```

Eine Code-Datei ausführen: `%run hello.py`

# Einige nützliche IPython-Features

| Feature | Beispiel |
|---------|----------|
| Hilfe / Dokumentation | `len?` |
| Code einfügen | `%paste` |
| Code-Laufzeit messen | `%timeit sum(range(1000))` |
| Code debuggen | `%debug` |
| Plots inline anzeigen | `%matplotlib inline` |

Mit `%` beginnende Befehle heißen **Magic Commands**.

# Methode 4: Jupyter Notebook

- Entwickelt für **Ju**lia, **Pyt**hon und **R** (daher **Ju**-**py**-te**R**)
- Bietet alle Funktionen der IPython-Shell
- Läuft im Browser
- Einfacheres Plotting
- Analysen lassen sich einfacher teilen (sogar Dashboards)
- *Extrem* beliebt, auch in Unternehmen

# Das Jupyter Notebook

## Zusätzliche Features

- Remote Execution
- Inline Plotting
- Interaktive Plots

## Nachteile

- Hidden State
- Versionskontrolle ist schwierig
- Testen ist schwierig

# Hidden State: ein Beispiel

Die Zellen unten laufen in der Reihenfolge, in der sie **ausgeführt** werden – nicht in der Reihenfolge, in der sie **stehen**.

````python
x = 5
````

````python
x = 100
````

````python
x * 3
````

Führst du die erste Zelle (`x = 5`) nach der zweiten erneut aus und danach die dritte, ergibt `x * 3` den Wert `15` statt `300` – obwohl im Notebook nichts anders aussieht.

**Tipp:** Vor dem Abgeben / Teilen: *Kernel → Restart & Run All*.

# Python-Skripte im Notebook ausführen

Mit `!` beginnende Zeilen führt das Notebook als Shell-Befehl aus:

````python
%%writefile hello.py
print("hello world")
````

````python
!python hello.py
````

# Pros und Cons der vier Methoden

| Methode | Pros | Cons |
|---------|------|------|
| **1** Skript per Shell | Versionskontrolle, saubere reproduzierbare Ausführung (kein Hidden State), präzise Zeitmessung | nicht interaktiv, Debugging schwieriger |
| **2** Python-Interpreter | interaktive Entwicklung | sehr unbequem |
| **3** IPython | interaktive Entwicklung / Debugging, Komfort, Dokumentation, Tab-Completion, History | Hidden State |
| **4** Jupyter Notebook | alle Vorteile von IPython, Inline Plotting, Code- und Markdown-Zellen für bessere Dokumentation | verleitet zu unstrukturiertem Code, Hidden State, Versionskontrolle schwieriger |
