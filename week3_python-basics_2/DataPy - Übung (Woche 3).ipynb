### Aufgabe A1: Bedingte List Comprehension

Schreibe `even_squares(numbers)`: Die Funktion bekommt eine Liste von Integers und gibt eine **neue** Liste mit den **Quadraten der geraden Zahlen** zurück. Verwende eine **List Comprehension**.

Beispiel: `[1, 2, 3, 4, 5, 6]` → `[4, 16, 36]`

````python
def even_squares(numbers: list[int]) -> list[int]:
    """Gibt die Quadrate der geraden Zahlen zurück."""
    # TODO

    
    return [n ** 2 for n in numbers if n % 2 == 0]
````

````python
data = [1, 2, 3, 4, 5, 6]
assert even_squares(data) == [4, 16, 36], "Beispiel aus der Aufgabe"
assert data == [1, 2, 3, 4, 5, 6], "Die Eingabeliste darf nicht verändert werden"
assert even_squares([]) == [], "Leere Liste"
assert even_squares([1, 3, 5]) == [], "Keine geraden Zahlen"
assert even_squares([-2, 0]) == [4, 0], "Negative Zahlen und 0"
print("✓ Aufgabe A1 bestanden")
````

````python
def add_tax(prices: list[float], rate: float) -> list[float]:
    """Berechnet die Bruttopreise (gerundet auf 2 Nachkommastellen)."""
    # TODO
    return list( map(lambda p: round(p * (1 + rate),2), prices) )
````

### Aufgabe A8: Generator-Funktion

Schreibe die **Generator-Funktion** `fibonacci(n)`: Sie liefert (mit `yield`) die **ersten `n` Fibonacci-Zahlen**, beginnend mit `0, 1, 1, 2, 3, 5, ...` (jede Zahl ist die Summe der beiden vorherigen). Für `n = 0` liefert sie nichts.

Beispiel: `list(fibonacci(7))` → `[0, 1, 1, 2, 3, 5, 8]`

*Hinweis: Die Funktion soll mit `yield` arbeiten und **nicht** eine fertige Liste zurückgeben.*

````python
def fibonacci(n: int) -> Iterator[int]:
    """Liefert die ersten n Fibonacci-Zahlen (0, 1, 1, 2, ...)."""
    # TODO
    a = 0
    b = 1
    for x in range(n):
       yield a
       temp = a
       a = b
       b = temp + b
````

````python
assert inspect.isgenerator(fibonacci(3)), "fibonacci muss ein Generator sein (yield statt return)"
assert list(fibonacci(7)) == [0, 1, 1, 2, 3, 5, 8], "Beispiel aus der Aufgabe"
assert list(fibonacci(0)) == [], "n = 0"
assert list(fibonacci(1)) == [0], "n = 1"
assert list(fibonacci(10))[-1] == 34, "Das 10. Element ist 34"
gen = fibonacci(3)
assert next(gen) == 0 and next(gen) == 1 and next(gen) == 1, "Werte einzeln mit next() abrufen"
print("✓ Aufgabe A8 bestanden")
````

````python
class Rectangle:
    """Ein Rechteck mit Breite und Höhe."""

    # TODO: __init__, area, __str__, __eq__
    def __init__(self, width, height):
        if(width < 0 or height < 0):
            raise ValueError("width or height cant be < 0!!")
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height
    
    def __str__(self):
        return f"Rectangle({self.width} x {self.height})"
    
    def __eq__(self, rect: Rectangle):
        return isinstance(rect, Rectangle) and self.width == rect.width and self.height == rect.height
       
````

### Aufgabe B2: Kapselung und eigene Exception

Definiere zuerst die **eigene Exception** `InsufficientFundsError` (sie erbt von `Exception`). Schreibe dann die Klasse `BankAccount` mit dem „privaten“ Attribut `__balance`:

- `__init__(self, balance=0)`: setzt den Anfangskontostand.
- `deposit(amount)`: zahlt `amount` ein. Ist `amount` nicht positiv, wird ein `ValueError` ausgelöst.
- `withdraw(amount)`: hebt `amount` ab. Ist nicht genug Guthaben vorhanden, wird `InsufficientFundsError` ausgelöst und der Kontostand **bleibt unverändert**.
- `get_balance()`: gibt den Kontostand zurück.

````python
class InsufficientFundsError(Exception):
    """Wird ausgelöst, wenn das Guthaben nicht reicht."""
    # TODO
    def __init__(self, balance):
        self.balance = balance
        super().__init__(f"Insufficient funds of only {balance} on your bank account!")

class BankAccount:
    """Ein Konto mit gekapseltem Kontostand."""

    # TODO: __init__, deposit, withdraw, get_balance
    def __init__(self, balance=0):
        self.__balance = balance
        

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("amount for deposit must be positive!")
        else:
            self.__balance += amount

    def withdraw(self, amount):
        if self.__balance - amount < 0:
            raise InsufficientFundsError(self.__balance)
        else:
            self.__balance -= amount

    def get_balance(self):
        return self.__balance
````

````python
assert issubclass(InsufficientFundsError, Exception), "Die Exception erbt von Exception"
account = BankAccount(100)
assert account.get_balance() == 100, "Anfangskontostand"
account.deposit(50)
assert account.get_balance() == 150, "deposit()"
account.withdraw(30)
assert account.get_balance() == 120, "withdraw()"
assert BankAccount().get_balance() == 0, "Standardwert 0"
assert_raises(InsufficientFundsError, account.withdraw, 500)
assert account.get_balance() == 120, "Bei einem Fehler bleibt der Kontostand unverändert"
assert_raises(ValueError, account.deposit, 0)
assert_raises(ValueError, account.deposit, -5)
assert hasattr(account, "_BankAccount__balance"), "Das Attribut soll __balance heißen (Name Mangling)"
assert not hasattr(account, "balance"), "Es soll kein öffentliches Attribut balance geben"
print("✓ Aufgabe B2 bestanden")
````

### Aufgabe B4: Eigener Iterator

Schreibe die Klasse `Countdown`, die eine **positive ganze Zahl `n`** bekommt und das **Iterator-Protokoll** (`__iter__`, `__next__`) implementiert. Beim Durchlaufen zählt sie von `n` bis `1` herunter. Danach wird `StopIteration` ausgelöst (dadurch endet eine `for`-Schleife).

Beispiel: `list(Countdown(3))` → `[3, 2, 1]`

````python
class Countdown:
    """Zählt von n bis 1 herunter."""

    # TODO: __init__, __iter__, __next__
    def __init__(self, n):
        self.n = n

    def __next__(self):
        if self.n <= 0:
            raise StopIteration
        
        self.n -= 1
        return self.n + 1
        

    def __iter__(self):
        return self
````

````python
assert list(Countdown(3)) == [3, 2, 1], "Beispiel aus der Aufgabe"
assert list(Countdown(1)) == [1], "n = 1"
assert list(Countdown(0)) == [], "n = 0 liefert nichts"
it = Countdown(2)
assert next(it) == 2 and next(it) == 1, "Werte einzeln mit next() abrufen"
assert_raises(StopIteration, next, it)
result = []
for number in Countdown(4):
    result.append(number)
assert result == [4, 3, 2, 1], "Verwendung in einer for-Schleife"
print("✓ Aufgabe B4 bestanden")
````

# Teil C: Hauptaufgabe "Notenverwaltung"

Ihr baut eine kleine Notenverwaltung. Für jede Person gibt es einen **Namen** und eine Liste von **Noten** (deutsche Skala: **1.0 ist am besten**, **4.0 ist gerade noch bestanden**, gültig sind Noten von 1.0 bis 5.0). Die Daten liegen in einer **Textdatei**, pro Zeile eine Person:

```
Alice;1.0,1.7,2.0
Dana;
```

Die Aufgabe besteht aus **sechs Schritten**, die aufeinander aufbauen: Die Klasse aus C1 verwendet ihr in allen späteren Schritten.

**Ziel:** aus einer Datei einen übersichtlichen Bericht erzeugen, der auch mit fehlenden Dateien und fehlerhaften Daten umgehen kann.

### Aufgabe C1: Die Klasse `Student`

Definiert die Exception `InvalidGradeError` (sie erbt von **`ValueError`**) und die Klasse `Student`:

- `__init__(self, name, grades=None)`: speichert `name` und eine **Kopie** der Notenliste in `self.grades`. Ohne Angabe ist die Liste leer. Liegt eine Note **außerhalb von 1.0 bis 5.0**, wird `InvalidGradeError` ausgelöst.
- `average()`: gibt den Durchschnitt der Noten zurück, bei **leerer** Notenliste `None`.
- `__str__()`: gibt `'Alice (Ø 1.57)'` (Durchschnitt mit 2 Nachkommastellen) zurück, ohne Noten `'Dana (keine Noten)'`.
- `from_line(line)`: eine **Klassenmethode** (`@classmethod`), die aus einer Textzeile der Form `'Alice;1.0,1.7,2.0'` ein `Student`-Objekt baut. Eine Zeile ohne Noten (`'Dana;'`) ergibt eine leere Notenliste.

*Tipps: `line.partition(';')` teilt an der ersten Stelle mit `;`. `float('1.7')` wandelt einen String in eine Zahl. Verwendet **keinen veränderlichen Standardwert** wie `grades=[]` im Funktionskopf.*

````python
class InvalidGradeError(ValueError):  # TODO: prüfen, ob die Vererbung stimmt
    """Wird ausgelöst, wenn eine Note ungültig ist."""
    pass

class Student:
    """Eine Person mit Namen und Noten."""

    # TODO: __init__, average, __str__ und die Klassenmethode from_line
    def __init__(self, name, grades=None):
        self.name = name
        self.grades = list(grades) if grades is not None else []
        for g in self.grades:
            if g < 1.0 or g > 5.0:
                raise InvalidGradeError

    def average(self):
        return None if not self.grades else sum(self.grades) / len(self.grades)

    def __str__(self):
        return f"{self.name} " + (f"(Ø {self.average():.2f})" if self.grades else "(keine Noten)")

    @classmethod
    def from_line(cls, line):
        name, semicolon, grades = line.partition(";")
        
        grades = [float(g) for g in grades.split(",")] if grades else []
        return cls(name, grades)
        
````

````python
assert issubclass(InvalidGradeError, ValueError), "InvalidGradeError erbt von ValueError"

alice = Student("Alice", [1.0, 1.7, 2.0])
assert alice.name == "Alice" and alice.grades == [1.0, 1.7, 2.0], "Attribute"
assert abs(alice.average() - 1.5667) < 0.001, "average()"
assert str(alice) == "Alice (Ø 1.57)", "__str__ mit Noten"

dana = Student("Dana")
assert dana.grades == [] and dana.average() is None, "Ohne Noten"
assert str(dana) == "Dana (keine Noten)", "__str__ ohne Noten"
assert Student("A").grades is not Student("B").grades, "Jede Person braucht ihre eigene Liste"

source = [2.0, 3.0]
copy_test = Student("X", source)
source.append(1.0)
assert copy_test.grades == [2.0, 3.0], "Die Notenliste soll kopiert werden"

assert_raises(InvalidGradeError, Student, "Y", [0.7])
assert_raises(InvalidGradeError, Student, "Y", [2.0, 5.3])
assert Student("Z", [1.0, 5.0]).average() == 3.0, "Die Grenzen 1.0 und 5.0 sind erlaubt"

parsed = Student.from_line("Alice;1.0,1.7,2.0")
assert isinstance(parsed, Student) and parsed.name == "Alice" and parsed.grades == [1.0, 1.7, 2.0], "from_line"
assert Student.from_line("Dana;").grades == [], "from_line ohne Noten"
assert Student.from_line("Chiara;1.3\n").grades == [1.3], "Zeilenumbruch am Ende"
assert_raises(InvalidGradeError, Student.from_line, "Q;9.0")
assert_raises(ValueError, Student.from_line, "Q;abc")
print("✓ Aufgabe C1 bestanden")
````

### Aufgabe C2: Dateien schreiben und lesen

Schreibt zwei Funktionen. Verwendet jeweils **`with open(...)`** und die Kodierung **`encoding='utf-8'`** (wegen Umlauten wie in „Jürgen“):

- `write_students(path, students)`: schreibt pro Person **eine Zeile** im Format `Name;note1,note2,...` (ohne Noten `Name;`).
- `read_students(path)`: liest die Datei und gibt eine Liste von `Student`-Objekten zurück. **Leere Zeilen** werden übersprungen. Verwendet `Student.from_line`.

````python
def write_students(path: str, students: list[Student]) -> None:
    """Schreibt pro Person eine Zeile 'Name;note1,note2' in die Datei."""
    # TODO
    
    with open(path, "w", encoding="utf-8") as f:
        for s in students:
            name = s.name
            grades = ",".join(str(g) for g in s.grades)
            f.write(f"{name};{grades}\n")


def read_students(path: str) -> list[Student]:
    """Liest die Personen aus der Datei (leere Zeilen werden übersprungen)."""
    # TODO
    with open(path, "r", encoding="utf-8") as f:
       return [ Student.from_line((line.strip())) for line in f if line.strip()] 
````

````python
import tempfile
from pathlib import Path

with tempfile.TemporaryDirectory() as folder:
    path = str(Path(folder) / "kurs.txt")
    write_students(path, students)

    text = Path(path).read_text(encoding="utf-8")
    lines = text.splitlines()
    assert len(lines) == len(students), "Eine Zeile pro Person"
    assert lines[0] == "Alice;1.0,1.7,2.0", f"Format der ersten Zeile: {lines[0]!r}"
    assert lines[4] == "Dana;", f"Person ohne Noten: {lines[4]!r}"
    assert "Jürgen" in text, "Umlaute müssen lesbar bleiben (UTF-8)"
    assert "Jürgen".encode("utf-8") in Path(path).read_bytes(), "Die Datei muss UTF-8-kodiert sein"

    loaded = read_students(path)
    assert [s.name for s in loaded] == [s.name for s in students], "Namen nach dem Einlesen"
    assert [s.grades for s in loaded] == [s.grades for s in students], "Noten nach dem Einlesen"
    assert all(isinstance(s, Student) for s in loaded), "Es sollen Student-Objekte sein"

    Path(path).write_text("Alice;1.0\n\nBob;2.0,3.0\n\n", encoding="utf-8")
    assert [s.name for s in read_students(path)] == ["Alice", "Bob"], "Leere Zeilen überspringen"
print("✓ Aufgabe C2 bestanden")
````

### Aufgabe C3: Wer hat bestanden?

Schreibe `passed_names(students)`: Die Funktion gibt die **Menge der Namen** aller Personen zurück, deren Durchschnitt **höchstens 4.0** beträgt. Personen **ohne Noten** haben nicht bestanden. Verwende eine **Set Comprehension**.

````python
def passed_names(students: list[Student]) -> set[str]:
    """Gibt die Namen aller Personen mit Durchschnitt <= 4.0 zurück."""
    # TODO
    return {s.name for s in students if s.grades and s.average() <= 4.0}
````

### Aufgabe C4: Ranking

Schreibe `ranking(students)`: Die Funktion gibt eine Liste von Tupeln `(name, durchschnitt)` zurück, **sortiert nach dem Durchschnitt aufsteigend** (beste Note zuerst). Der Durchschnitt wird auf **2 Nachkommastellen gerundet**. Personen **ohne Noten** kommen nicht vor. Die Eingabeliste bleibt unverändert.

*Tipp: `sorted(..., key=lambda item: item[1])` sortiert nach dem zweiten Element jedes Tupels.*

````python
def ranking(students: list[Student]) -> list[tuple[str, float]]:
    """Gibt (Name, Durchschnitt) sortiert nach Durchschnitt zurück."""
    # TODO
    with_grades = [(s.name, round(s.average(), 2)) for s in students if s.grades]
    
    return sorted(with_grades, key=lambda x: x[1])
````

````python
expected = [("Chiara", 1.3), ("Alice", 1.57), ("Bob", 2.65), ("Jürgen", 4.23), ("Eli", 5.0)]
before = [s.name for s in students]
assert ranking(students) == expected, f"Ergebnis: {ranking(students)}"
assert [s.name for s in students] == before, "Die Eingabeliste darf nicht verändert werden"
assert ranking([]) == [], "Leere Liste"
assert ranking([Student("Dana")]) == [], "Nur Personen ohne Noten"
print("✓ Aufgabe C4 bestanden")
````

### Aufgabe C5: Zeilen für den Bericht: ein Generator

Schreibe die **Generator-Funktion** `grade_lines(ranked)`: Sie bekommt das Ergebnis von `ranking(...)` und liefert (mit `yield`) pro Eintrag **eine formatierte Zeile**: Platz, Name (linksbündig auf Breite 8) und Durchschnitt mit 2 Nachkommastellen. Die Plätze beginnen bei 1.

Beispiel: Für `[('Chiara', 1.3), ('Alice', 1.57)]` liefert sie `'1. Chiara   1.30'` und `'2. Alice    1.57'`.

*Tipps: `enumerate(..., start=1)` zählt die Plätze mit. Im f-String richtet `{name:<8}` den Namen linksbündig auf Breite 8 aus, `{average:.2f}` formatiert auf 2 Nachkommastellen.*

````python
def grade_lines(ranked: list[tuple[str, float]]) -> Iterator[str]:
    """Liefert pro Eintrag eine Zeile 'Platz. Name Durchschnitt'."""
    # TODO
    for rank, student in enumerate(ranked, start=1):
        yield f"{rank}. {student[0]:<8} {student[1]:.2f}"
````

````python
lines = grade_lines([("Chiara", 1.3), ("Alice", 1.57)])
assert inspect.isgenerator(lines), "grade_lines muss ein Generator sein (yield)"
assert next(lines) == "1. Chiara   1.30", "Erste Zeile"
assert next(lines) == "2. Alice    1.57", "Zweite Zeile"
assert_raises(StopIteration, next, lines)
assert list(grade_lines([])) == [], "Leeres Ranking"
assert list(grade_lines([("Jürgen", 4.2)])) == ["1. Jürgen   4.20"], "Zwei Nachkommastellen"
print("✓ Aufgabe C5 bestanden")
````

### Aufgabe C6: Alles zusammen: der Bericht

Schreibe `load_report(path)`: Die Funktion liest die Datei mit `read_students`, bildet das `ranking` und gibt alle Zeilen aus `grade_lines` **durch Zeilenumbrüche (`'\n'`) getrennt** als **einen String** zurück.

Fehler fangt ihr mit **`try`/`except`** ab und gebt stattdessen einen Text zurück:

- Datei existiert nicht (`FileNotFoundError`): `"Fehler: Datei '<path>' nicht gefunden."`
- Daten sind ungültig (`ValueError`, dazu gehört auch `InvalidGradeError`): `'Fehler: ungültige Daten (<Fehlermeldung>).'`

*Tipp: Mit `except ValueError as error:` bekommt ihr die Fehlermeldung als `error`.*

````python
def load_report(path: str) -> str:
    """Liest die Datei und gibt den Bericht als String zurück (oder eine Fehlermeldung)."""
    # TODO
    try:
        students = read_students(path)
    except FileNotFoundError:
        return f"Fehler: Datei '{path}' nicht gefunden."
    except ValueError as error:
        return f"Fehler: ungültige Daten ({error})"

    return "\n".join(grade_lines(ranking(students)))
````
