# Tipps zur Suche

- **Beschreibung** lesen: Wer hat die Daten wann und wie erhoben?
- Auf **Spaltenbeschreibung** (Data Dictionary) achten
- **Zwei Kandidaten** suchen, dann entscheiden

# Ist der Datensatz geeignet? (1/2)

- **Umfang:** genug Zeilen (z. B. mehrere hundert) und Spalten, aber nicht zu groß für euren Laptop
- **Spaltentypen:** Mischung aus **numerischen** und **kategorialen** Spalten
- **Dokumentation:** Bedeutung der Spalten ist klar
- **Qualität:** nicht zu viele fehlende Werte
- **Aktualität:** passt zur Fragestellung
- **Datenschutz:** keine Daten, mit denen sich Personen identifizieren lassen
- **Synthetische Daten:** nur mit Begründung, denn die Ergebnisse sagen oft wenig über die Realität aus
- **Lizenz** erlaubt die Nutzung, Quelle ist angegeben
- Bei ML: **Zielvariable** vorhanden, Klassen nicht extrem unausgewogen

# Lizenzen: kurz erklärt

| Lizenz | Bedeutung |
|---|---|

# Einlesen: richtige Parameter

```python
df = pd.read_csv(
    "data/zaehlstellen.csv",
    sep=";",                 # Trennzeichen
    decimal=",",             # Dezimalkomma
    encoding="latin-1",      # Kodierung
    parse_dates=["Datum"],
    date_format="%d.%m.%Y",
)
df.head(5)
```

```python
df.dtypes
```

# Eine wiederverwendbare Prüffunktion

```python
def quick_check(df, target=None):
    """Erste Datenprüfung eines DataFrames."""
    print("Form:", df.shape)
    print("\nDatentypen:")
    print(df.dtypes.to_string())
    missing = df.isna().sum()
    print("\nFehlende Werte:")
    if missing.any():
        print(missing[missing > 0].to_string())
    else:
        print("keine")
    print("\nDuplikate:", df.duplicated().sum())
    if target is not None:
        print(f"\nVerteilung von '{target}':")
        share = df[target].value_counts(normalize=True)
        print(share.round(2).to_string())
```

```python
quick_check(df, target="Wetter")
```

```python
df.describe()
```

Die Maximalwerte fallen auf: **99999** bei `Anzahl` ist sehr wahrscheinlich ein Platzhalter oder Fehler. Wie ihr mit **fehlenden Werten, Duplikaten und Ausreißern** umgeht, lernen wir in **Woche 7**.
