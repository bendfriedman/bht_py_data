### Aufgabe A1: Liste filtern und verdoppeln

Schreibe `manipulate_list(lst)`: Die Funktion bekommt eine Liste von Integers und gibt eine **neue** Liste zurück, die nur die **geraden** Zahlen enthält, jeweils **verdoppelt**.

Beispiel: `[1, 2, 3, 4, 5]` → `[4, 8]`

````python
def manipulate_list(lst: list[int]) -> list[int]:
    """Gibt die geraden Zahlen aus lst zurück, jeweils verdoppelt."""
    # TODO
    newLst = []
    for el in lst:
        if el % 2 == 0:
            newLst.append(el * 2)
    return newLst
````

````python
data = [1, 2, 3, 4, 5]
assert manipulate_list(data) == [4, 8], "Beispiel aus der Aufgabe"
assert data == [1, 2, 3, 4, 5], "Die Eingabeliste darf nicht verändert werden"
assert manipulate_list([]) == [], "Leere Liste"
assert manipulate_list([1, 3]) == [], "Keine geraden Zahlen"
print("✓ Aufgabe A1 bestanden")
````

### Aufgabe A2: Dictionary umkehren

Schreibe `swap_dict(d)`: Die Funktion gibt ein **neues** Dictionary zurück, in dem Schlüssel und Werte vertauscht sind.

Beispiel: `{'a': 1, 'b': 2}` → `{1: 'a', 2: 'b'}`

*Annahme: Die Werte sind eindeutig und als Schlüssel zulässig.*

````python
def swap_dict(d: dict) -> dict:
    """Vertauscht Schlüssel und Werte."""
    # TODO
    newDict = {}
    for k, v in d.items():
        newDict[v] = k
    return newDict
````

````python
original = {'a': 1, 'b': 2}
assert swap_dict(original) == {1: 'a', 2: 'b'}, "Beispiel aus der Aufgabe"
assert original == {'a': 1, 'b': 2}, "Das Original darf nicht verändert werden"
assert swap_dict({}) == {}, "Leeres Dictionary"
print("✓ Aufgabe A2 bestanden")
````

### Aufgabe A6: Durchschnittsnoten

Schreibe `get_student_grades(students)`: Die Funktion bekommt eine **Liste von Dictionaries** mit Namen und Noten und gibt eine **Liste von Tupeln** `(Name, Durchschnittsnote)` zurück.

Beispiel: `[{'name': 'Alice', 'grades': [90, 80]}, {'name': 'Bob', 'grades': [70, 75, 80]}]` → `[('Alice', 85.0), ('Bob', 75.0)]`

````python
def get_student_grades(students: list[dict]) -> list[tuple[str, float]]:
    """Gibt für jede Person (Name, Durchschnitt der grades) zurück."""
    # TODO
    result = []
    for student in students:
        name = student['name']
        grades = student['grades']
        avgGrade = sum(grades) / len(grades)
        result.append((name, avgGrade))
    return result
    
        
````

### Aufgabe B3: Countdown mit while-Schleife

Schreibe `countdown(n)`: Die Funktion zählt mit einer **while-Schleife** von der positiven ganzen Zahl `n` bis `0` herunter und **gibt jede Zahl in einer eigenen Zeile aus**.

Ist `n` **keine positive ganze Zahl**, gibt sie stattdessen genau diese Fehlermeldung aus: `Fehler: n muss eine positive ganze Zahl sein.`

Beispiel: `countdown(3)` gibt `3`, `2`, `1`, `0` aus (jeweils in einer Zeile).

````python
def countdown(n: int) -> None:
    """Zählt von n bis 0 herunter."""
    # TODO
    if not isinstance(n, int) or n <= 0:
        print("Fehler: n muss eine positive ganze Zahl sein.")
    else:
        i = n
        while i >= 0:
            print(i)
            i -= 1
````

# Teil C: Hauptaufgabe "Kursauswertung"

Ihr wertet die Ergebnisse eines (fiktiven) Kurses aus. Für jede Person liegt vor:

- `grades`: Liste der Noten (deutsche Skala: **1.0 ist am besten**, **4.0 ist gerade noch bestanden**),

Die Aufgabe besteht aus **sieben Schritten**, die aufeinander aufbauen: Die Funktionen aus den ersten Schritten verwendet ihr in den späteren.

**Ziel:** am Ende einen übersichtlichen Bericht ausgeben, wer wie abgeschnitten hat und wer bestanden hat.

### Die Daten

Führt die Zelle aus. Der Datensatz wird in allen Schritten verwendet.

````python
students = [
    {"name": "Alice",  "grades": [1.3, 2.0, 1.7],      "project": True},
    {"name": "Bram",   "grades": [3.0, 2.7],           "project": True},
    {"name": "Chiara", "grades": [1.0, 1.3, 1.0, 1.7], "project": True},
    {"name": "Deniz",  "grades": [4.0, 3.7, 4.3],      "project": False},
    {"name": "Emil",   "grades": [5.0, 4.7],           "project": True},
    {"name": "Fenja",  "grades": [],                   "project": False},
    {"name": "Gül",    "grades": [2.3, 2.7, 2.0],      "project": True},
]
````

### Aufgabe C1: average

**Schritt 1: Durchschnitt berechnen**

Schreibe `average(values)`: Die Funktion gibt den **Durchschnitt** einer Liste von Noten zurück, **auf 2 Nachkommastellen gerundet** (`round`). Ist die Liste leer, gibt sie `None` zurück.

````python
def average(values: list[float]) -> float | None:
    """Durchschnitt auf 2 Nachkommastellen, None bei leerer Liste."""
    # TODO
    if not values:
        return None

    sum = 0 
    for v in values:
        sum += v
    return round(sum / len(values), 2)
````

````python
assert average([1.3, 2.0, 1.7]) == 1.67
assert average([3.0, 2.7]) == 2.85
assert average([4.0]) == 4.0
assert average([]) is None, "Leere Liste → None"
print("✓ Schritt 1 bestanden")
````

### Aufgabe C2: grade_label

**Schritt 2: Note in Worte fassen**

Schreibe `grade_label(avg)` mit einer **if-elif-else-Kette**. Die Funktion bekommt einen (gerundeten) Durchschnitt und gibt zurück:

| Durchschnitt | Rückgabe |
|---|---|
| bis 1.5 | `'sehr gut'` |
| bis 2.5 | `'gut'` |
| bis 3.5 | `'befriedigend'` |
| bis 4.0 | `'ausreichend'` |
| über 4.0 | `'nicht bestanden'` |
| `None` | `'keine Note'` |

*Achtung: Prüfe zuerst auf `None`, bevor du Zahlen vergleichst.*

````python
def grade_label(avg: float | None) -> str:
    """Wandelt einen Notendurchschnitt in eine Bewertung um."""
    # TODO
    if not avg:
        return "keine Note"
    elif ( avg <= 1.5 ):
        return ("sehr gut")
    elif (avg <= 2.5 ):
        return ("gut")
    elif( avg <= 3.5 ):
        return ("befriedigend")
    elif( avg <= 4.0):
        return ("ausreichend")
    else:
        return ("nicht bestanden")
````

````python
expected = {1.0: 'sehr gut', 1.5: 'sehr gut', 1.67: 'gut', 2.5: 'gut', 2.85: 'befriedigend',
            3.5: 'befriedigend', 4.0: 'ausreichend', 4.01: 'nicht bestanden', 5.0: 'nicht bestanden'}
for avg, label in expected.items():
    assert grade_label(avg) == label, f"grade_label({avg}) sollte '{label}' liefern"
assert grade_label(None) == 'keine Note'
print("✓ Schritt 2 bestanden")
````

### Aufgabe C3: evaluate

**Schritt 3: Alle auswerten**

Schreibe `evaluate(students)`: Die Funktion geht mit einer **for-Schleife** durch die Liste (Datensatz siehe oben) und gibt eine **Liste von Tupeln** `(name, avg, label)` zurück. Nutze dafür `average` und `grade_label`.

Beispiel für Alice: `('Alice', 1.67, 'gut')`

````python
def evaluate(students: list[dict]) -> list[tuple[str, float | None, str]]:
    """Gibt für jede Person (name, avg, label) zurück."""
    # TODO
    tupel_list = []
    for s in students:  
        name = s["name"]
        avg = average(s["grades"])
        label = grade_label(avg)
        tupel_list.append((name, avg, label))
    return tupel_list


    # students = [
    #     {"name": "Alice",  "grades": [1.3, 2.0, 1.7],      "project": True},
    #     {"name": "Bram",   "grades": [3.0, 2.7],           "project": True},
    #     {"name": "Chiara", "grades": [1.0, 1.3, 1.0, 1.7], "project": True},
    #     {"name": "Deniz",  "grades": [4.0, 3.7, 4.3],      "project": False},
    #     {"name": "Emil",   "grades": [5.0, 4.7],           "project": True},
    #     {"name": "Fenja",  "grades": [],                   "project": False},
    #     {"name": "Gül",    "grades": [2.3, 2.7, 2.0],      "project": True},
    # ]
````

````python
results = evaluate(students)
assert len(results) == 7
assert results[0] == ('Alice', 1.67, 'gut')
assert results[3] == ('Deniz', 4.0, 'ausreichend')
assert results[5] == ('Fenja', None, 'keine Note')
assert results[6] == ('Gül', 2.33, 'gut')
print("✓ Schritt 3 bestanden")
````

### Aufgabe C4: ranking

**Schritt 4: Rangliste**

Schreibe `ranking(results)`: Die Funktion gibt die Ergebnisse **nach Durchschnitt aufsteigend** sortiert zurück (beste Note zuerst). Personen **ohne Note** (`None`) stehen **am Ende**. Die Eingabeliste darf nicht verändert werden.

*Tipp: Teile die Liste mit einer Schleife in zwei Listen (mit und ohne Note). Sortiere die erste Liste mit `sorted(..., key=...)`, wobei `key` eine kleine Hilfsfunktion ist, die den Durchschnitt aus einem Tupel herausgibt.*

````python
def ranking(results: list[tuple]) -> list[tuple]:
    """Sortiert nach Durchschnitt, Personen ohne Note am Ende."""
    # TODO
    s_with_grades = [r for r in results if r[1] is not None]
    s_no_grades = [r for r in results if r[1] is None]

    return sorted(s_with_grades, key=lambda x: x[1]) + s_no_grades
````

````python
original = evaluate(students)
ranked = ranking(original)
names = [entry[0] for entry in ranked]
assert names == ['Chiara', 'Alice', 'Gül', 'Bram', 'Deniz', 'Emil', 'Fenja'], names
assert original == evaluate(students), "Die Eingabeliste darf nicht verändert werden"
print("✓ Schritt 4 bestanden")
````

### Aufgabe C5: Sets

**Schritt 5: Wer hat bestanden?**

Schreibe `passed_students(students)`. Eine Person hat bestanden, wenn sie **ein Projekt abgegeben** hat **und** ihre Bewertung weder `'nicht bestanden'` noch `'keine Note'` ist.

Die Funktion gibt ein Tupel aus **drei Sets** von Namen zurück:

1. alle, die **bestanden** haben,
2. alle, die **nur eine** der beiden Bedingungen erfüllen (Projekt **oder** Note, nicht beides),
3. alle, die **keine** der beiden Bedingungen erfüllen.

*Tipp: Bilde zuerst zwei Sets (Projekt abgegeben, Note ausreichend) und nutze dann Mengenoperationen (`&`, `^`).*

````python
def passed_students(students: list[dict]) -> tuple[set, set, set]:
    """Gibt (bestanden, nur_eine_bedingung, keine_bedingung) als Sets von Namen zurück."""
    # TODO
    projekt_abgegeben: set = {s["name"] for s in students if s["project"]}
    note_ausreichend: set = {s["name"] for s in students if grade_label(average(s["grades"])) not in ("nicht bestanden","keine Note")}
    all_names: set = {s["name"] for s in students}

    passed = projekt_abgegeben & note_ausreichend
    only_one = projekt_abgegeben ^ note_ausreichend
    either = projekt_abgegeben | note_ausreichend
    failed = all_names - either

    return (passed, only_one, failed)


     # students = [
        #     {"name": "Alice",  "grades": [1.3, 2.0, 1.7],      "project": True},
        #     {"name": "Bram",   "grades": [3.0, 2.7],           "project": True},
        #     {"name": "Chiara", "grades": [1.0, 1.3, 1.0, 1.7], "project": True},
        #     {"name": "Deniz",  "grades": [4.0, 3.7, 4.3],      "project": False},
        #     {"name": "Emil",   "grades": [5.0, 4.7],           "project": True},
        #     {"name": "Fenja",  "grades": [],                   "project": False},
        #     {"name": "Gül",    "grades": [2.3, 2.7, 2.0],      "project": True},
        # ]
````

````python
passed, only_one, neither = passed_students(students)
assert passed == {'Alice', 'Bram', 'Chiara', 'Gül'}, passed
assert only_one == {'Deniz', 'Emil'}, only_one
assert neither == {'Fenja'}, neither
print("✓ Schritt 5 bestanden")
````

### Aufgabe C6: count_labels

**Schritt 6: Verteilung der Bewertungen**

````python
def count_labels(results: list[tuple]) -> dict[str, int]:
    """Zählt, wie oft jede Bewertung vorkommt."""
    # TODO
    count_l = {}
    for r in results:
        label = grade_label(r[1])
        count_l[label] = count_l.get(label, 0) + 1

    return count_l
````

### Aufgabe C7: format_report

**Schritt 7: Bericht mit f-Strings**

Schreibe `format_report(ranked)`: Die Funktion bekommt die Rangliste aus Schritt 4 und gibt **einen mehrzeiligen String** zurück:

- eine **Kopfzeile** (`Rang`, `Name`, `Ø`, `Bewertung`),
- eine **Trennlinie** aus `-`,
- danach **eine Zeile pro Person** mit Rang, Name, Durchschnitt (2 Nachkommastellen, bei `None` ein `-`) und Bewertung.

Gestalte die Spalten mit **Format-Spezifikationen** im f-String (z. B. `{name:<8}` für linksbündig, `{avg:>5.2f}` für rechtsbündig mit 2 Nachkommastellen). Eine mögliche Ausgabe (gekürzt):

```
Rang  Name         Ø  Bewertung
--------------------------------
   1  Chiara    1.25  sehr gut
   2  Alice     1.67  gut
   ...
   7  Fenja        -  keine Note
```

*Die genaue Breite der Spalten ist dir überlassen, die Reihenfolge der Werte nicht.*

````python
def format_report(ranked: list[tuple]) -> str:
    """Erzeugt eine Tabelle als mehrzeiligen String."""
    # TODO
    result = []
    result.append(f"{"Rang": <5} {"Name": <8} {"Ø": >6} {"Bewertung"} ")
    result.append("-" * 35)
    for i, s in enumerate(ranked):
        rank = i+1
        name = s[0]
        grade = f"{s[1]:.2f}" if s[1] is not None else "-"
        label = s[2]
        
        result.append(f"{rank: <5} {name: <8} {grade: >6} {label} ")

    return "\n".join(result)
````
