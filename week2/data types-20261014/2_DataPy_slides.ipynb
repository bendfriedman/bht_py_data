# PEP 8: Der Python-Style-Guide

- **PEP** = ***P**ython **E**nhancement **P**roposal*
- PEP 8 beschreibt, wie Python-Code **aussehen** sollte
- Grundgedanke: Code wird **öfter gelesen als geschrieben** („Readability counts“)
- Einheitlicher Stil erleichtert die Zusammenarbeit

Offizielles Dokument: [peps.python.org/pep-0008](https://peps.python.org/pep-0008/)

# PEP 8: Namenskonventionen

| Was | Konvention | Beispiel |
|-----|------------|----------|
| Variablen, Funktionen | `snake_case` | `mean_price`, `add_numbers()` |
| Konstanten | `UPPER_CASE` | `MAX_ROWS` |
| Klassen | `CapWords` | `DataLoader` |
| Module, Packages | kurze, kleingeschriebene Namen | `utils`, `datenanalyse` |

- Aussagekräftige Namen verwenden: `age_in_years` statt `a`
- Variablennamen müssen mit Buchstabe oder `_` beginnen und sind **case sensitive**

# PEP 8: Layout

- **4 Leerzeichen** pro Einrückungsebene (keine Tabs mischen)
- Zeilenlänge: höchstens **79 Zeichen**
- Zwei Leerzeilen zwischen Funktionen auf oberster Ebene
- Imports am Dateianfang, ein Import pro Zeile
- Leerzeichen um Operatoren, nicht aber bei Keyword-Argumenten

```python
# Nicht empfohlen
def f(a,b = 2):
  return a+b

# Empfohlen
def add(a, b=2):
    return a + b
```

# Variablen sind Zeiger: Beispiel

- **Java:** Eine Variable eines primitiven Typs *ist* der Container, die Zuweisung überschreibt den Wert
- **Python:** Eine Variable ist ein Name, der auf ein Objekt im Speicher zeigt

````python
x = 3.1415   # Objekt mit Wert 3.1415 anlegen, Name x zeigt darauf
y = x        # zweiter Name y zeigt auf dasselbe Objekt
print(x)
print(y)
print(id(x))  # Adresse des Objekts, auf das x zeigt
print(id(y))
````

# Variablen sind Zeiger: Neuzuweisung

````python
x = 2.71     # neues Objekt mit Wert 2.71, x zeigt jetzt darauf
print(x)
print(y)
print(id(x))
print(id(y))
````

Java würde den Wert im selben Container von 3.1415 auf 2.71 überschreiben.
Python legt dagegen **zwei Objekte** an (3.1415 und 2.71) und lässt nur den Zeiger `x` umziehen.

# Variablen sind Zeiger: Bonus

````python
# auch x += 3 erzeugt ein neues Objekt (2.71 + 3 = 5.71)
x += 3
print(x)
print(id(x))
````

**Merke:** Python unterscheidet **unveränderliche** (immutable: `int`, `float`, `bool`, `str`, `tuple`) und **veränderliche** (mutable: `list`, `dict`, `set`) Objekte. Dazu gleich mehr.

# Arithmetische Operatoren

| Operator | Name | Beschreibung |
|----------|------|--------------|
| `a + b` | Addition | Summe von a und b |
| `a - b` | Subtraktion | Differenz von a und b |
| `a * b` | Multiplikation | Produkt von a und b |
| `a / b` | echte Division (true division) | Quotient von a und b |
| `a // b` | ganzzahlige Division (floor division) | Quotient von a und b ohne Nachkommastellen |
| `a % b` | Modulo | Rest der ganzzahligen Division von a durch b |
| `a ** b` | Potenz | a hoch b |
| `-a` | Negation | das Negative von a |
| `+a` | unäres Plus | a unverändert (selten genutzt) |

# Arithmetische Operatoren: Beispiele

````python
a = 12
b = 3
print(a + b)
print(a / b, a // b, a % b, a ** b)
print(-7 // 2)   # floor division rundet nach unten ab
# Operatorrangfolge (Punkt-vor-Strich)
print(a + b / 3 - 4 * 12)
````

# Bitweise Operatoren

| Operator | Name | Beschreibung |
|----------|------|--------------|
| `a & b` | Bitwise AND | Bits, die in a und b gesetzt sind |
| `a \| b` | Bitwise OR | Bits, die in a oder b oder beiden gesetzt sind |
| `a ^ b` | Bitwise XOR | Bits, die in a oder b, aber nicht in beiden gesetzt sind |
| `a << b` | Bit Shift links | Bits von a um b Stellen nach links schieben |
| `a >> b` | Bit Shift rechts | Bits von a um b Stellen nach rechts schieben |
| `~a` | Bitwise NOT | bitweise Negation von a |

````python
a = True
b = False
print(a & b)
x, y = 5, 3          # binär: 101 und 011
print(x & y, x | y, x ^ y, x << 1, x >> 1, ~x)
````

# Zuweisungsoperatoren

Für jeden arithmetischen und bitweisen Operator gibt es eine erweiterte Zuweisung:

| | | |
|---|---|---|
| `a += b` | `a -= b` | `a *= b` |
| `a //= b` | `a %= b` | `a **= b` |
| `a \|= b` | `a ^= b` | `a <<= b` |

`a += 2` entspricht `a = a + 2`, wertet `a` aber nur einmal aus.

````python
a = 2
print(id(a))
a += 2
print(a)
print(id(a))   # neue Adresse: Integers sind unveränderlich
````

# Boolesche Operatoren

| Operator | Beschreibung |
|----------|--------------|
| `a and b` | wahr, wenn a und b wahr sind |
| `a or b` | wahr, wenn a oder b wahr ist |
| `not a` | wahr, wenn a falsch ist |

````python
print(True and False)
print([True, True] or [False, True])
print([True, True, True] and [True, False, 2])
print(0 or 'Standardwert')
````

`and` und `or` liefern bei nicht-booleschen Werten **einen der beiden Operanden** zurück:

- `a or b` liefert `a`, falls `a` wahr ist, sonst `b`
- `a and b` liefert `a`, falls `a` falsch ist, sonst `b`

# Vergleichsoperatoren

| Operator | Beschreibung | Operator | Beschreibung |
|----------|--------------|----------|--------------|
| `a == b` | a gleich b | `a != b` | a ungleich b |
| `a < b` | a kleiner als b | `a > b` | a größer als b |
| `a <= b` | a kleiner oder gleich b | `a >= b` | a größer oder gleich b |

````python
# ist 25 ist gerade?
print(25 % 2 == 0)
# verkettete Vergleiche: liegt a zwischen 15 und 30?
a = 25
print(15 < a < 30)
# Vergleich von Listen liefert ein einzelnes bool (nicht elementweise)
print([2, 2] == [2, 2])
````

# Identitäts- und Zugehörigkeitsoperatoren

| Operator | Beschreibung |
|----------|--------------|
| `a is b` | wahr, wenn a und b **dasselbe Objekt** sind |
| `a is not b` | wahr, wenn a und b nicht dasselbe Objekt sind |
| `a in b` | wahr, wenn a Element von b ist |
| `a not in b` | wahr, wenn a nicht Element von b ist |

# Identität und Zugehörigkeit: Beispiele

````python
a = [1, 2]
b = a
print(a == b)    # gleicher Inhalt
print(a is b)    # dasselbe Objekt (gleiche Adresse)
c = [1, 2]
print(c == a)    # gleicher Inhalt
print(c is a)    # aber ein anderes Objekt
print(1 in [1, 2, 3])
print([3] in [3, [3]])
````

# Zeiger auf Listen: Referenz oder Kopie?

# Szenario 1: Verketten mit `+`

````python
a = [1, 2, 3, 5]    # Listen-Container anlegen, Zeiger a zeigt darauf
b = a               # b zeigt auf denselben Container
a = a + [8]         # neuer Container wird angelegt, a zeigt jetzt darauf
print(a)            # a zeigt auf den neuen, erweiterten Container
print(b)            # b zeigt weiter auf den ursprünglichen Container
````

Listen mit `+` zu verketten erzeugt eine **neue** Liste.

# Szenario 1.1: erweiterte Zuweisung

````python
# Szenario 1.1: erweiterte Zuweisung
a = [1, 2, 3, 5]
b = a
a += [8]            # der ursprüngliche Container wird in-place verändert
print(a)
print(b)            # b sieht die Änderung, denn es ist derselbe Container
````

# Szenario 2: Listenmethode

````python
# Szenario 2: Listenmethode
a = [1, 2, 3, 5]
b = a
a.append(8)         # derselbe Container wird verändert
print(a)
print(b)
````

Listenmethoden (`append`, `pop`, `insert`, ...) erzeugen **keinen** neuen Container, sondern verändern den bestehenden.

# Szenario 3: Element über `b` ändern

````python
# Szenario 3: Änderung über b
a = [1, 2, 3, 5]
b = a
b[0] = 7            # erstes Element im gemeinsamen Container ändern
print(a)
print(b)
````

# Szenario 4: Kopie mit `[:]`

````python
# Szenario 4: Kopie mit [:]
a = [1, 2, 3, 5]
b = a[:]            # !!! Kopie des Containers, b zeigt auf den neuen Container
b[0] = 7            # ändert nur den zweiten Container
print(a)
print(b)
````

`[ ]` greift direkt auf denselben Container zu. Ein leeres Slicing `[:]` erzeugt dagegen eine **Kopie**.

# Flache vs. tiefe Kopie

Eine Kopie mit `[:]` oder `.copy()` ist **flach** (shallow): Verschachtelte Listen werden nicht mitkopiert.

````python
import copy

a = [[1, 2], [3, 4]]
b = a.copy()              # flache Kopie
b[0].append(99)           # verändert die innere Liste, die a und b teilen
print(a)
c = copy.deepcopy(a)      # tiefe Kopie: auch die inneren Listen werden kopiert
c[0].append(100)
print(a)
print(c)
````

# Sets

- **ungeordnete** Sammlungen **eindeutiger** Elemente
- unterstützen Mengenoperationen
- definiert durch `{1, 2, 3}`

````python
primes = {2, 3, 5, 7}
odds = {1, 3, 5, 7, 9}
print(primes)
print({1, 2, 2, 3, 3, 3})   # Duplikate werden entfernt
````

# Mengenoperationen

| Operation | Bedeutung | Operator | Methode |
|-----------|-----------|----------|---------|
| Vereinigung (Union) | Elemente in einer der Mengen | `a \| b` | `a.union(b)` |
| Schnittmenge (Intersection) | Elemente in beiden Mengen | `a & b` | `a.intersection(b)` |
| Differenz (Difference) | Elemente in a, aber nicht in b | `a - b` | `a.difference(b)` |
| Symmetrische Differenz | Elemente in genau einer Menge | `a ^ b` | `a.symmetric_difference(b)` |

````python
print(primes | odds, primes.union(odds))
print(primes & odds, primes.intersection(odds))
print(primes - odds, primes.difference(odds))
print(primes ^ odds, primes.symmetric_difference(odds))
````

# Type Hints

Python ist **dynamisch typisiert**: Typen müssen nicht deklariert werden.

Type Hints (Typannotationen) ermöglichen:

- bessere Lesbarkeit des Codes
- bessere Autovervollständigung und Fehlererkennung in der IDE
- statische Typprüfung (mit Tools wie `mypy`)
- Dokumentation

**Wichtig:** Type Hints sind optional und haben **keinen Einfluss auf die Laufzeit**. Python bleibt dynamisch typisiert.

# Type Hints: Beispiel

````python
# Ohne Type Hints
def greet(name):
    return f"Hello, {name}!"

# Mit Type Hints
def greet_typed(name: str) -> str:
    return f"Hello, {name}!"

print(greet("Alice"))
print(greet_typed("Bob"))

# Type Hints verhindern nicht, dass das läuft (mypy würde es aber melden)
print(greet_typed(123))
````

# Bedingte Anweisungen: if-elif-else

````python
x = 1e6

if x == 0:
    print(x, "is zero")
elif x > 0:                  # darf nur nach einem 'if' stehen
    print(x, "is positive")
elif x < 0:
    print(x, "is negative")
else:                        # darf nur nach einem 'if' oder 'elif' stehen
    print(x, "is unlike anything I've ever seen before...")
````

# Verschachtelte Schleifen: Performance

Jede weitere verschachtelte Schleife multipliziert die Rechenzeit mit **N** (Anzahl der Durchläufe).

**Gute Praxis:** unnötige Schleifen vermeiden, z. B. durch **Vektorisierung**, also Funktionen aus Drittbibliotheken, die auf ganzen Vektoren arbeiten (mehr dazu bei NumPy und Pandas).

# break vs. continue

Manchmal soll der vorgegebene Ablauf einer Schleife unterbrochen werden:

````python
students_grades = {"Freddy": 1.3, "Laura": 1.0, "Bobby": 5.0, "Betty": 2.3}
everyone_passed = True

for stud in students_grades:
    if students_grades[stud] > 4.0:
        everyone_passed = False
        break    # beendet die for-Schleife, weiter geht es nach der Schleife

if everyone_passed:
    print("Good job.")
else:
    print("Someone didn't pass")
````

# break vs. continue: continue

````python
for stud in students_grades:
    if students_grades[stud] > 4.0:
        continue   # bricht den aktuellen Durchlauf ab, weiter mit dem nächsten

    print(f"Good job, {stud}")
````

# while-Schleifen

Wiederholen, **bis eine Bedingung erfüllt ist**. Vorsicht vor Endlosschleifen!

````python
i = 0
while i < 10:
    print(i, end=' ')
    i += 1
````

- Oft lässt sich eine for-Schleife in eine while-Schleife übersetzen und umgekehrt
- **Faustregel:** Ist die (maximale) Anzahl der Durchläufe vorab bekannt, nimm eine for-Schleife, sonst eine while-Schleife

# Funktionen definieren

````python
def add(a, b):
    """
    Addiert zwei Zahlen.

    Input
    a: eine Zahl
    b: eine weitere Zahl

    Returns Summe von a und b
    """
    result = a + b
    return result


add(1, 1)
````

- `def` leitet die Funktionsdefinition ein, `return` gibt einen Wert zurück
- Der **Docstring** (`"""..."""`) dokumentiert die Funktion, siehe `help(add)`

# Funktionen mit mehreren Parametern

````python
def add_and_print(a, b, print_result):
    """
    Addiert zwei Zahlen.

    Input
    a: eine Zahl
    b: eine weitere Zahl
    print_result: True, falls Ergebnis ausgegeben wird

    Returns Summe von a und b
    """
    result = a + b
    if print_result:
        print(f"Your result is {result}")
    return result


add_and_print(1, 1, True)
````

# Python übergibt Referenzen

Argumente werden als **Referenz auf das Objekt** übergeben. Ändert die Funktion nur ihren *Namen* (Neuzuweisung), bleibt das Objekt außerhalb unverändert:

````python
def add_ten(x):
    print('Inside function, before adding: ' + str(x))
    x = x + 10
    print('Inside function, after  adding: ' + str(x))


x = 5
print('Before function: ' + str(x))
add_ten(x)
print('After function:  ' + str(x))
````

# Rückgabewerte nutzen

````python
def add_ten_and_return(x):
    print('Inside function, before adding: ' + str(x))
    x = x + 10
    print('Inside function, after  adding: ' + str(x))

    return x


x = 5
print('Before function: ' + str(x))
x = add_ten_and_return(x)
print('After function:  ' + str(x))
````

# Veränderliche Objekte als Argumente

Bei **veränderlichen** Objekten (z. B. Listen) wirkt sich eine Änderung *im* Objekt auch außerhalb der Funktion aus:

````python
def append_ten(lst):
    lst.append(10)      # verändert das Objekt selbst


numbers = [1, 2, 3]
append_ten(numbers)
print(numbers)
````
